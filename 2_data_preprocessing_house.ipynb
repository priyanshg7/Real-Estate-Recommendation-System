```python
import re  # Import Python's regular expression module for pattern-based text cleaning

# Clean the 'society' column by removing rating symbols like "4★" or "4.5 ★"
df['society'] = (
    df['society']
    
    # Apply a lambda function to each society name
    .apply(
        lambda name: 
        # re.sub() replaces matched patterns with an empty string
        # Pattern explanation:
        # \d+        -> matches one or more digits (e.g., 4, 12)
        # (\.\d+)?  -> optionally matches decimal part (e.g., .5 in 4.5)
        # \s?       -> optionally matches a space before the star
        # ★          -> matches the star symbol used for ratings
        # Result: removes patterns like "4★", "4.5★", "3 ★" from society names
        re.sub(r'\d+(\.\d+)?\s?★', '', str(name))
        
        # Remove leading and trailing spaces after cleaning
        .strip()
    )
    
    # Convert all society names to lowercase for consistency
    .str.lower()
)
```

```python
# For rows where the price is mentioned in 'Lac',
# convert the value to 'Crore' by dividing by 100
# (since 1 Crore = 100 Lacs)
df.loc[
    df['price'].str.contains('Lac'),  # select rows where price unit is 'Lac'
    'price_num'                       # target numeric price column
] /= 100                              # convert Lacs → Crores
```
